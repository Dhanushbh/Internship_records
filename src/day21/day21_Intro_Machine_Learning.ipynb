```python
# day18_Intro_machine_learning.py

import pandas as pd
from sklearn.model_selection import train_test_split

# Load dataset
df = pd.read_csv("Bangalore.csv")

# Basic inspection
print(df.head())
print(df.info())

# Feature-label separation (example: predict price)
X = df.drop(columns=['Price'])   # change 'price' to your target column name
y = df['Price']

# Train-test split (70-30 with random_state)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```
